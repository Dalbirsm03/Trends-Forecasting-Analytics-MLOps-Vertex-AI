```python
import pandas as pd
```

```python
file_path = r"C:/Users/Dalbir/Downloads/Trends-Forecasting-Analytics-MLOps-Vertex-AI/data/processed_file/sales_cleaned.parquet"
df = pd.read_parquet(file_path)
```

```python
df[["year", "week", "day"]] = df["order_date"].dt.isocalendar()
```

```python
df["day_name"] = df['order_date'].dt.day_name()
```

```python
df["weekday_weekend"] = df["day_name"].map(lambda x: "Weekend" if x in ["Saturday", "Sunday"] else "Weekday")
```

```python
df["order_size"] = df["quantity_ordered"].map(
    lambda x: "single" if x == 1 else ("multiple" if x <= 5 else "bulk")
)
```

```python
df['quarter'] = pd.to_datetime(df['order_date']).dt.quarter
```

```python
df["day_name"] = df["day_name"].astype("string")
df["weekday_weekend"] = df["weekday_weekend"].astype("string")
df["order_size"] = df["order_size"].astype("string")
```

```python
df["3_day_shift_ma"] = df["sales"].rolling(3).mean().shift(1)
```
