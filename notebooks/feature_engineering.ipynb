```python
import pandas as pd
```

```python
file_path = r"C:/Users/Dalbir/Downloads/Trends-Forecasting-Analytics-MLOps-Vertex-AI/data/processed_file/sales_cleaned.parquet"
df = pd.read_parquet(file_path)
```

```python
df[["year", "week", "day"]] = df["order_date"].dt.isocalendar()
```

```python
df["day_name"] = df['order_date'].dt.day_name()
```

```python
df["weekday_weekend"] = df["day_name"].map(lambda x: "Weekend" if x in ["Saturday", "Sunday"] else "Weekday")
```

```python
df["order_size"] = df["quantity_ordered"].map(
    lambda x: "single" if x == 1 else ("multiple" if x <= 5 else "bulk")
)
```

```python
df['quarter'] = pd.to_datetime(df['order_date']).dt.quarter
```

```python
df["day_name"] = df["day_name"].astype("string")
df["weekday_weekend"] = df["weekday_weekend"].astype("string")
df["order_size"] = df["order_size"].astype("string")
```

```python
df['SMA_3'] = df['sales'].rolling(window=3).mean()
df['SMA_5'] = df['sales'].rolling(window=5).mean()
```

```python
df['Lag_3'] = df['sales'].shift(3)
df['Lag_5'] = df["sales"].shift(5)
```

```python
df["CMA"] = df["sales"].expanding().mean()
```

```python
df.info()
```

```python
df[['SMA_3','SMA_5','Lag_3','Lag_5']] = df[['SMA_3','SMA_5','Lag_3','Lag_5']].fillna(0)
```

```python
output_path = r"C:/Users/Dalbir/Downloads/Trends-Forecasting-Analytics-MLOps-Vertex-AI/data/processed_file/sales_featured.parquet"
df.to_parquet(output_path, index=False)
print("Feature-engineered data saved successfully!")
```

```python
import pandas as pd
from statsmodels.tsa.arima.model import ARIMA
import matplotlib.pyplot as plt

# Only the target column is needed
y = df["sales"]

# Fit ARIMA model (order = p,d,q)
model = ARIMA(y, order=(1,1,1))
model_fit = model.fit()

# Forecast next 3 months
forecast = model_fit.forecast(steps=3)
print(forecast)

# Plot
plt.plot(y, label="Actual")
plt.plot(forecast, label="Forecast", color="red")
plt.legend()
plt.show()
```
